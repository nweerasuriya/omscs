# Question 1

```python
lower, upper = 0, 1
def pdf(x):
    return 3 * x**2

# find expectation and variance of x if x is distributed according to the pdf above
x = np.linspace(lower, upper, 100)
expectation = np.trapz(x * pdf(x), x)
variance = np.trapz((x - expectation)**2 * pdf(x), x)
print(f"Expectation: {expectation}")
print(f"Variance: {variance}")
```

```python
def pdf(x):
    return np.sin(x)

lower, upper = 0, math.pi/2
x = np.linspace(lower, upper, 100)
# find expectation and variance of x if x is distributed according to the pdf above
expectation = np.trapz(x * pdf(x), x)
variance = np.trapz((x - expectation)**2 * pdf(x), x)
print(f"Expectation: {expectation}")
print(f"Variance: {variance}")
```
